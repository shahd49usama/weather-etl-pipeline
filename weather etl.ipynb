```python
from datetime import datetime
list_cities = [] 
ingestion_time = datetime.now()
Country = 'EG'

for city in cities:
   
    params = {
    'appid' : '6a13e61e853763292101962a026aa3e8',
    "q": f"{city},EG"    }

    response = requests.get(url, params=params)
    data = response.json()


    Latitude = data['coord']['lat']
    Longitude = data['coord']['lon']
    WeatherMain = data['weather'][0]['main']
    WeatherDescription = data['weather'][0]['description']
    Temperature = data['main']['temp']
    Pressure = data['main']['pressure']
    Humidity  = data['main']['humidity']
    WindSpeed = data['wind']['speed']
    TempMin = data['main']['temp_min']
    TempMax = data['main']['temp_max']
    WindDeg = data['wind']['deg']
    Cloudiness = data['clouds']['all']
    Visibility = data['visibility']
    Sunrise = data['sys']['sunrise']
    Sunset = data['sys']['sunset']
    

    list_cities.append({
        'City': city,
        'Country': Country,
        'Latitude': Latitude,
        'Longitude': Longitude,
        'Temperature': Temperature,
        'TempMin': TempMin,
        'TempMax': TempMax,
        'Pressure': Pressure,
        'Humidity': Humidity,
        'Cloudiness': Cloudiness,
        'Visibility': Visibility,
        'WeatherMain': WeatherMain,
        'WeatherDescription': WeatherDescription,
        'WindSpeed': WindSpeed,
        'WindDeg': WindDeg,
        'Sunrise': Sunrise,
        'Sunset': Sunset,
        'IngestionTime': ingestion_time
    
    })
```

```python
from sqlalchemy import create_engine

engine = create_engine(
    "mssql+pyodbc://localhost/weather_dwh2"
    "?trusted_connection=yes"
    "&driver=ODBC+Driver+17+for+SQL+Server"
    "&TrustServerCertificate=yes"
)
```

```python
df = pd.read_sql("""
    SELECT *
    FROM bronze.Bronze_WeatherData
    WHERE IngestionTime = (SELECT MAX(IngestionTime) FROM bronze.Bronze_WeatherData)
""", con=engine)
```

```python
for col in ['Temperature', 'TempMin', 'TempMax']:
    df[col] = (df[col] - 273.15).round(2)

df[['City', 'Temperature', 'TempMin', 'TempMax']]
```

```python
df['City'] = df['City'].str.strip().str.title()
df['WeatherDescription'] = df['WeatherDescription'].str.strip().str.title()

region_map = {
    'Cairo': 'Greater Cairo',
    'Giza': 'Greater Cairo',
    'Alexandria': 'North Coast',
    'Port Said': 'Canal',
    'Suez': 'Canal',
    'Ismailia': 'Canal',
    'Mansoura': 'Lower Egypt',
    'Tanta': 'Lower Egypt',
    'Luxor': 'Upper Egypt',
    'Aswan': 'Upper Egypt'
}
df['Region'] = df['City'].map(region_map).fillna('Unknown')

df[['City', 'Region']]
```

```python
df['TemperatureCategory'] = np.select(
    [df['Temperature'] < 15, df['Temperature'] <= 30],
    ['Cold', 'Moderate'],
    default='Hot'
)

df['HumidityCategory'] = np.select(
    [df['Humidity'] < 40, df['Humidity'] <= 70],
    ['Low', 'Medium'],
    default='High'
)

df['CloudCategory'] = np.select(
    [df['Cloudiness'] <= 25, df['Cloudiness'] <= 70],
    ['Clear', 'Partly Cloudy'],
    default='Cloudy'
)

df[['City', 'Temperature', 'TemperatureCategory', 'Humidity', 'HumidityCategory']]
```

```python
df = df.rename(columns={'WindSpeed': 'WindSpeed_ms'})
df['WindSpeed_kmh'] = (df['WindSpeed_ms'] * 3.6).round(2)
df['Visibility_km'] = (df['Visibility'] / 1000).round(2)

directions = ['North','North_East','East','South_East','South','South_West','West','North_West']
df['WindDirection'] = df['WindDeg'].apply(
    lambda d: directions[int(((d + 22.5) % 360) // 45)] if pd.notnull(d) else None
)

df[['City', 'WindSpeed_ms', 'WindSpeed_kmh', 'WindDeg', 'WindDirection', 'Visibility_km']]
```

```python
for col in ['Sunrise', 'Sunset']:
    df[col] = (pd.to_datetime(df[col], unit='s', utc=True)
                 .dt.tz_convert('Africa/Cairo')
                 .dt.tz_localize(None))

df['DaylightHours'] = ((df['Sunset'] - df['Sunrise']).dt.total_seconds() / 3600).round(2)

df[['City', 'Sunrise', 'Sunset', 'DaylightHours']]
```

```python
df['LoadTime'] = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
```

```python
silver_cols = ['City','Country','Latitude','Longitude','Temperature','TempMin',
               'TempMax','Pressure','Humidity','Cloudiness','CloudCategory',
               'Visibility_km','WindSpeed_ms','WindSpeed_kmh','WindDeg',
               'WindDirection','WeatherMain','WeatherDescription','Sunrise',
               'Sunset','DaylightHours','Region','TemperatureCategory',
               'HumidityCategory','IngestionTime']

df_silver = df[silver_cols]
```

```python
df_silver.to_sql(
    "Silver_WeatherData",     
    engine,
    schema="silver",
    if_exists="append",
    index=False
)
```

```python
import pandas as pd
import numpy as np
from sqlalchemy import create_engine, text

engine = create_engine(
    "mssql+pyodbc://localhost/weather_dwh2"
    "?trusted_connection=yes"
    "&driver=ODBC+Driver+17+for+SQL+Server"
    "&TrustServerCertificate=yes"
)

df = pd.read_sql("""
    SELECT *
    FROM silver.Silver_WeatherData
    WHERE IngestionTime = (SELECT MAX(IngestionTime) FROM silver.Silver_WeatherData)
""", con=engine)
```

```python
df['TempRange'] = (df['TempMax'] - df['TempMin']).round(2)
print(df[['City', 'TempMin', 'TempMax', 'TempRange']])
```

```python
T = df['Temperature']
H = df['Humidity']

df['HeatIndex'] = (
    T + 0.33 * (H / 100 * 6.105 * np.exp(17.27 * T / (237.7 + T))) - 4.0
).round(2)

print(df[['City', 'Temperature', 'Humidity', 'HeatIndex']])
```

```python
df['WindCategory'] = np.select(
    [df['WindSpeed_kmh'] < 12, df['WindSpeed_kmh'] <= 30],
    ['Calm', 'Breezy'],
    default='Windy'
)
print(df[['City', 'WindSpeed_kmh', 'WindCategory']])
```

```python
temp_penalty = (df['Temperature'] - 24).abs() * 4
hum_penalty  = (df['Humidity'] - 50).abs() * 0.5

df['ComfortScore'] = (100 - temp_penalty - hum_penalty).clip(0, 100).round(1)

df['ComfortCategory'] = np.select(
    [df['ComfortScore'] >= 75, df['ComfortScore'] >= 50],
    ['Comfortable', 'Acceptable'],
    default='Uncomfortable'
)
print(df[['City', 'Temperature', 'Humidity', 'ComfortScore', 'ComfortCategory']])
```

```python
df['TempRank'] = df['Temperature'].rank(ascending=False, method='min').astype(int)
print(df[['City', 'Temperature', 'TempRank']].sort_values('TempRank'))
```

```python
df['WeatherDate'] = df['IngestionTime'].dt.date
df['WeatherHour'] = df['IngestionTime'].dt.hour
```

```python
gold_cols = ['City','Country','Latitude','Longitude','Temperature','TempMin',
             'TempMax','TempRange','HeatIndex','Pressure','Humidity','Cloudiness',
             'CloudCategory','Visibility_km','WindSpeed_ms','WindSpeed_kmh',
             'WindCategory','WindDeg','WindDirection','WeatherMain',
             'WeatherDescription','Sunrise','Sunset','DaylightHours',
             'Region','TemperatureCategory','HumidityCategory','ComfortScore',
             'ComfortCategory','TempRank','WeatherDate','WeatherHour',
             'IngestionTime']

df_gold = df[gold_cols]
```

```python
with engine.begin() as conn:
    conn.execute(text("TRUNCATE TABLE gold.Gold_WeatherData"))

df_gold.to_sql("Gold_WeatherData", engine, schema="gold",
               if_exists="append", index=False)
```
